Finance Assistant – Pseudocode
START PROGRAM
OPEN application

CREATE Tkinter GUI window

ASK user to enter their name
STORE user's name

DISPLAY Main Menu

Main Menu:
    1. Add CSV Financial Data
    2. Dashboard
    3. Ask Chatbot
    4. Exit


IF user selects "Add CSV Financial Data":

    OPEN CSV Data page

    DISPLAY required CSV data format

    ASK user to select a CSV file

    READ CSV file using Pandas

    CLEAN and organise the financial data

    CHECK that the CSV contains the required information

    IF the data is valid:

        STORE the financial data

        DISPLAY "Data successfully added"

    ELSE:

        DISPLAY "Invalid CSV data"

    ASK:
        "Would you like to add another CSV?"

    IF YES:

        RETURN to CSV Data page

        ADD new data to existing data

    ELSE:

        RETURN to Main Menu


IF user selects "Dashboard":

    IF no financial data has been uploaded:

        DISPLAY "Please upload CSV data first"

    ELSE:

        OPEN Dashboard

        CALCULATE:
            Total Current Amount
            Total Income
            Total Expenses

        DISPLAY the three totals

        CREATE Bar Chart for Spending by Date

        CREATE Bar Chart for Spending by Category

        DISPLAY both charts using Matplotlib

        CREATE Date Filter

        CREATE Type/Category Filter

        IF user changes a filter:

            FILTER the financial data

            RECALCULATE the totals

            UPDATE the charts

            DISPLAY updated results

        RETURN to Main Menu


IF user selects "Ask Chatbot":

    IF no financial data has been uploaded:

        DISPLAY "Please upload CSV data first"

    ELSE:

        OPEN Chatbot page

        DISPLAY text box

        ASK user to enter a question

        READ user's question

        ANALYSE the financial data

        IF question asks for most expensive expense:

            FIND expense with the largest amount

            DISPLAY the expense and amount

        ELSE IF question asks for cheapest expense:

            FIND expense with the smallest amount

            DISPLAY the expense and amount

        ELSE IF question asks for total income:

            CALCULATE total income

            DISPLAY total income

        ELSE IF question asks for total expenses:

            CALCULATE total expenses

            DISPLAY total expenses

        ELSE IF question asks about spending by category:

            GROUP expenses by category

            CALCULATE spending for each category

            DISPLAY the results

        ELSE:

            DISPLAY "I cannot answer this question yet"

        ASK user if they want to ask another question

        IF YES:

            ASK another question

        ELSE:

            RETURN to Main Menu


IF user selects "Exit":

    CLOSE Tkinter application

    END PROGRAM

END PROGRAM
